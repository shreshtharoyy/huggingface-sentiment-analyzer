```python
import transformers, torch
print("Transformers version:", transformers.__version__)
print("PyTorch version:", torch.__version__)
```

```python
from transformers import pipeline
classifier = pipeline("sentiment-analysis")
```

```python
while True:
    text = input("Enter text (or type 'exit'): ")

    if text.lower() == "exit":
        print("Exiting Sentiment Analyzer")
        break

    result = classifier(text)

    label = result[0]["label"]
    score = result[0]["score"]

    print(f"Sentiment: {label}")
    print(f"Confidence: {score * 100:.2f}%")
```

```python
sample_reviews = [
    "Worst experience ever.",
    "The movie was decent, not great.",
    "I really enjoyed using this application.",
    "I would never recommend this to anyone."
]

for review in sample_reviews:
  result = classifier(review)

  label = result[0]["label"]
  score = result[0]["score"]

  print(f"\nReview: {review}")
  print(f"Label: {label}")
  print(f"Confidence: {score * 100:.2f}%")
```
